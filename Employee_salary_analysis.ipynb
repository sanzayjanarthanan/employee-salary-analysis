```python
import pandas as pd

data = {
    "Employee": ["Arun", "Priya", "Karthik", "Divya", "Rahul", "Meena"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"],
    "Salary": [45000, 38000, 52000, 41000, 60000, 36000],
}
df = pd.DataFrame(data)

total = df["Salary"].sum()
average = df["Salary"].mean()
highest = df["Salary"].max()
lowest = df["Salary"].min()

top = df.loc[df["Salary"].idxmax(), "Employee"]
bottom = df.loc[df["Salary"].idxmin(), "Employee"]

print(f"Total salary   : {total}")
print(f"Average salary : {average:.2f}")
print(f"Highest salary : {highest} ({top})")
print(f"Lowest salary  : {lowest} ({bottom})")

# Verify against original data (hint from task)
assert total == sum(data["Salary"])
assert round(average, 2) == round(sum(data["Salary"]) / len(data["Salary"]), 2)

# Extra: department-wise average
print(df.groupby("Department")["Salary"].mean())
```
